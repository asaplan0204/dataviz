````python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt

docs = [
    "the model trains on data and the data shapes the model",
    "visualization turns data into pictures people can read",
    "a topic model finds themes hiding in a corpus of text",
    "embeddings place similar text near similar text in space",
    "people read pictures faster than they read tables of data",
    "text needs a transform before any visualization can begin",
    "a corpus of documents becomes a matrix of token counts",
    "the matrix view of text loses word order but gains structure",
]
print(len(docs), 'documents')
````

## Part 1 — Counts: the honest view vs. the poster

````python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(stop_words="english")
counts = cv.fit_transform(docs)
freq = pd.Series(counts.sum(axis=0).A1, index=cv.get_feature_names_out())

from wordcloud import WordCloud
wc = WordCloud(width=420, height=280, background_color="white",
               colormap="GnBu").generate_from_frequencies(freq.to_dict())

fig, axes = plt.subplots(1, 2, figsize=(11, 3.6))
freq.sort_values().tail(10).plot.barh(ax=axes[0], color="#2E6E8E")
axes[0].set_title("Sorted bars: is #3 bigger than #5? Instant.")
axes[1].imshow(wc); axes[1].axis("off")
axes[1].set_title("Word cloud: same data — now rank #3 vs #5")
plt.tight_layout(); plt.show()
````

**Run the Week 2 test live:** ask the room to rank terms 3–5 from the cloud, then reveal the bars. The gap *is* the area-vs-position ranking.

## Part 2 — TF–IDF: frequent → distinctive

````python
from sklearn.feature_extraction.text import TfidfVectorizer

tfv = TfidfVectorizer(stop_words="english")
T = tfv.fit_transform(docs)

d = 3   # the embeddings doc
raw = pd.Series(counts[d].toarray()[0], index=cv.get_feature_names_out())
tfd = pd.Series(T[d].toarray()[0], index=tfv.get_feature_names_out())

fig, axes = plt.subplots(1, 2, figsize=(10.5, 3.4))
raw[raw > 0].sort_values().plot.barh(ax=axes[0], color="#5a6672")
axes[0].set_title("Doc 4, RAW counts")
tfd[tfd > 0].sort_values().plot.barh(ax=axes[1], color="#2E6E8E")
axes[1].set_title("Same doc, TF–IDF: corpus-common words sink")
plt.tight_layout(); plt.show()
````

## Part 5 — BERTopic's middle step, by hand

BERTopic is four stages: **embed → project (UMAP) → cluster (HDBSCAN) → describe (c-TF-IDF)**.
Parts 1–4 covered counting, weighting and projecting. The clustering stage is the
one students take on trust, so run it directly — on a corpus where the right
answer is known, so the methods can be judged.

````python
# A synthetic corpus: four clear topics of unequal size, plus 60 off-topic
# documents, which is what a real corpus looks like.
rng = np.random.default_rng(0)
TOPICS = {
    "sport":   "match team player season goal coach league score injury training stadium".split(),
    "finance": "market stock price investor bank rate profit trading shares fund yield".split(),
    "health":  "patient trial drug clinic dose symptom therapy doctor study disease risk".split(),
    "tech":    "model data network chip software compute device memory kernel latency cloud".split(),
}
ALL = sum(TOPICS.values(), [])

syn, truth = [], []
for ti, (name, pool) in enumerate(TOPICS.items()):
    for _ in range([140, 100, 60, 40][ti]):
        syn.append(" ".join(list(rng.choice(pool, 12)) + list(rng.choice(ALL, 3))))
        truth.append(ti)
for _ in range(60):                       # belong to no topic
    syn.append(" ".join(rng.choice(ALL, 15))); truth.append(4)
truth = np.array(truth)

Xs = TfidfVectorizer().fit_transform(syn).toarray()
print(Xs.shape, "documents x terms")
````

````python
# Project first. Clustering a 2-D projection is exactly what BERTopic does.
import umap
Y = umap.UMAP(n_components=2, random_state=0).fit_transform(Xs)

try:
    from sklearn.cluster import HDBSCAN          # scikit-learn >= 1.3
except ImportError:
    from hdbscan import HDBSCAN                  # the standalone package
from sklearn.cluster import KMeans, DBSCAN

runs = [
    ("true topics",     truth),
    ("KMeans, k=4",     KMeans(4, n_init=10, random_state=0).fit_predict(Y)),
    ("KMeans, k=7",     KMeans(7, n_init=10, random_state=0).fit_predict(Y)),
    ("DBSCAN, eps=0.4", DBSCAN(eps=0.4, min_samples=8).fit_predict(Y)),
    ("DBSCAN, eps=0.8", DBSCAN(eps=0.8, min_samples=8).fit_predict(Y)),
    ("HDBSCAN",         HDBSCAN(min_cluster_size=20).fit_predict(Y)),
]

palette = np.array(["#2E6E8E", "#d9534f", "#6aa84f", "#b07aa1", "#e6a23c",
                    "#4f7f5f", "#a0522d"])
fig, axes = plt.subplots(2, 3, figsize=(12, 6))
for ax, (name, lab) in zip(axes.ravel(), runs):
    colours = ["#cccccc" if v == -1 else palette[v % len(palette)] for v in lab]
    found = len(set(lab) - {-1})
    noise = int((np.asarray(lab) == -1).sum())
    ax.scatter(Y[:, 0], Y[:, 1], c=colours, s=8)
    ax.set_title(f"{name}  ({found} groups, {noise} unassigned)", fontsize=10)
    ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.show()
````

**Narrate the three failures and the one success:**

- **KMeans must be told `k`.** At `k=4` it is right; at `k=7` it splits real topics
  apart, and it looks just as confident either way. It also assigns every
  off-topic document to some cluster, because it has no way to say "none".
- **DBSCAN** can say "none" — grey points — but the answer moves with `eps`.
  Compare the two panels: same data, same code, one parameter.
- **HDBSCAN** varies the density threshold across the space, so it chooses the
  number of groups itself. That is why BERTopic uses it, and why BERTopic has a
  **topic −1**: the documents it declined to assign.
